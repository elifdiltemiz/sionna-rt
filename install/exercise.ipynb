```python
import sionna.rt
from sionna.rt import (
    load_scene,
    Transmitter,
    Receiver,
    PlanarArray,
    PathSolver,
    Camera
)

# 1. Load Munich scene
scene = load_scene(sionna.rt.scene.munich)

# 2. Set frequency
scene.frequency = 3.5e9

# 3. Configure TX antenna
scene.tx_array = PlanarArray(
    num_rows=1,
    num_cols=1,
    pattern="iso",
    polarization="V"
)

# 4. Configure RX antenna
scene.rx_array = PlanarArray(
    num_rows=1,
    num_cols=1,
    pattern="iso",
    polarization="V"
)

# 5. Add transmitter
tx = Transmitter(
    name="TX",
    position=[8.5, 21, 27]
)
scene.add(tx)

# 6. Add receiver
rx = Receiver(
    name="RX",
    position=[45, 90, 1.5]
)
scene.add(rx)

# 7. Calculate radio propagation paths
solver = PathSolver()
paths = solver(scene, max_depth=3)
```

```python
# 8. Show TX, RX and radio propagation paths
from google.colab import output
output.enable_custom_widget_manager()
scene.preview(paths=paths)
```
